```python
# Load env variables
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
from anthropic import Anthropic

anthropic_client = Anthropic()
anthropic_model = "claude-sonnet-4-0"
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)
```

```python
def chat(messages):
    # Make a request using the Anthropic SDK
    message = anthropic_client.messages.create(
        model=anthropic_model,
        max_tokens=1000,
        messages=messages,
    )
    return message.content[0].text
```

```python
anthropicMessages = []

add_user_message(anthropicMessages, "What is quantum computing? Answer in one sentence")

answer = chat(anthropicMessages)

answer
```

```python
import os
from openai import OpenAI

# Initialize client pointing to AICredits gateway
openai_client = OpenAI(
    base_url="https://api.aicredits.in/v1",
    api_key=os.getenv("AICREDITS_API_KEY")
)

model = "anthropic/claude-haiku-4.5" 

# Helper functions to manage message history
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages):
    response = openai_client.chat.completions.create(
        model=model,
        max_tokens=1000,
        messages=messages,
    )
    # OpenAI SDK returns text in response.choices[0].message.content
    return response.choices[0].message.content
```

```python
messages = []

# Turn 1
add_user_message(messages, "What is quantum computing? Answer in one sentence")
answer = chat(messages)
add_assistant_message(messages, answer)
print("Assistant:", answer)

# Turn 2 (context preserved)
add_user_message(messages, "Write another sentence explaining qubits.")
final_answer = chat(messages)
add_assistant_message(messages, final_answer)
print("\nAssistant:", final_answer)
```
