````python
import os
from dotenv import load_dotenv

load_dotenv()
````

````python
def add_user_message(messages, text):
    messages.append({"role": "user", "content": text})


def add_assistant_message(messages, text):
    messages.append({"role": "assistant", "content": text})
````

````python
import json
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

openai_client = OpenAI(
    base_url="https://api.aicredits.in/v1",
    api_key=os.getenv("AICREDITS_API_KEY"),
)
openai_model = "anthropic/claude-sonnet-4-0"


def openai_chat(messages, system=None, stop=None):
    chat_messages = []
    if system:
        chat_messages.append({"role": "system", "content": system})
    chat_messages.extend(messages)

    response = openai_client.chat.completions.create(
        model=openai_model,
        messages=chat_messages,
        max_tokens=1000,
        stop=stop,
    )
    return response.choices[0].message.content


# Usage
messages = []
add_user_message(messages, "Generate a very short event bridge rule as json")
add_assistant_message(messages, "```json")

text = openai_chat(messages, stop=["```"])

# Parse the JSON string
data = json.loads(text.strip())
print(data)
````

````python
import json
import os
from anthropic import Anthropic
from dotenv import load_dotenv

load_dotenv()

anthropic_client = Anthropic(api_key=os.getenv("ANTHROPIC_API_KEY"))
anthropic_model = "claude-3-7-sonnet-latest"


def anthropic_chat(messages, system=None, stop_sequences=None):
    params = {
        "model": anthropic_model,
        "max_tokens": 1000,
        "messages": messages,
    }
    if system:
        params["system"] = system
    if stop_sequences:
        params["stop_sequences"] = stop_sequences

    message = anthropic_client.messages.create(**params)
    return message.content[0].text


# Usage
messages = []
add_user_message(messages, "Generate a very short event bridge rule as json")
add_assistant_message(messages, "```json")

text = anthropic_chat(messages, stop_sequences=["```"])

# Parse the JSON string
data = json.loads(text.strip())
print(data)
````
