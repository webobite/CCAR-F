```python
messages = []
add_user_message(messages, "Write a 1 sentence description of a fake database")

# Request a streaming response
stream = openai_client.chat.completions.create(
    model=model,  # e.g., "anthropic/claude-sonnet-4-0"
    messages=messages,
    max_tokens=1000,
    stream=True,
)

collected_text = []

for chunk in stream:
    content = chunk.choices[0].delta.content
    if content:
        print(content, end="", flush=True)
        collected_text.append(content)

# Equivalent to stream.get_final_message()
final_message = "".join(collected_text)
print("\n\nFinal assembled output:\n", final_message)
```

```python
from anthropic import Anthropic

anthropic_client = Anthropic()
anthropic_model = "claude-sonnet-4-0"

def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None):
    params = {
        "model":anthropic_model,
        "max_tokens":1000,
        "messages":messages,
    }
    if system:
        params["system"] = system

        
    # Make a request using the Anthropic SDK
    message = anthropic_client.messages.create(**params)
    return message.content[0].text
```
