```python
import os
from openai import OpenAI

openai_client = OpenAI(
    base_url="https://api.aicredits.in/v1",
    api_key=os.getenv("AICREDITS_API_KEY"),
)
model = "anthropic/claude-sonnet-4-0"


def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)


def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)


def chat(messages, system=None):
    # In OpenAI, the system prompt is passed as the first message in the list
    chat_messages = []
    if system:
        chat_messages.append({"role": "system", "content": system})
    chat_messages.extend(messages)

    # Make request using the OpenAI SDK
    response = openai_client.chat.completions.create(
        model=model,
        max_tokens=1000,
        messages=chat_messages,
    )
    return response.choices[0].message.content
```

```python
messages = []
system = """
You are a patient math tutor. Do not directly answer a student's questions. 
Guide them to a solution step by step.
"""

# Add student message
add_user_message(messages, "How do I solve 5x + 3 = 2 for x?")

# Call the chat function
answer = chat(messages, system=system)
print(answer)

# If continuing the conversation, record the assistant's reply and the next step
add_assistant_message(messages, answer)
add_user_message(messages, "Should I subtract 3 from both sides?")

follow_up_answer = chat(messages, system=system)
print(follow_up_answer)
```

```python
from anthropic import Anthropic

anthropic_client = Anthropic()
anthropic_model = "claude-sonnet-4-0"

def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None):
    params = {
        "model"=anthropic_model,
        "max_tokens"=1000,
        "messages"=messages,
    }
    if system:
        params["system"] = system

        
    # Make a request using the Anthropic SDK
    message = anthropic_client.messages.create(**params)
    return message.content[0].text
```
