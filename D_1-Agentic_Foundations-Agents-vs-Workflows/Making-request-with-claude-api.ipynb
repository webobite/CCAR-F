```python
# Load env variables
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
# Anthropic import and model defined
from anthropic import Anthropic
client = Anthropic()
model = "claude-sonnet-4-0"
```

```python
# Make a request
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence"
        }
    ]
)
```

```python
from openai import OpenAI
# Initialize OpenAI client pointed at AICredits gateway
client = OpenAI(
    base_url="https://api.aicredits.in/v1",
    api_key=os.getenv("AICREDITS_API_KEY")  # Or paste "sk-..." directly
)

# You can use OpenAI models or Claude models routed via AICredits
model = "anthropic/claude-haiku-4.5"  # or "anthropic/claude-3-5-sonnet"

# Make the chat completion request
response = client.chat.completions.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence"
        }
    ]
)

# Extract and print the generated text
print(response.choices[0].message.content)
```
